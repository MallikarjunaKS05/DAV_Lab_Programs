```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("students.csv")
```

```python
print(df)
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(
    x=df["Marks"]
)
plt.title("Box plot of student Marks")
plt.xlabel("Marks")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(
    df["Marks"],
    bins=10
)
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    x="StudyHours",
    y="Marks",
    data=df,
    hue="Result",
    palette={
        "Pass":"green",
        "Fail":"red"
    },
    s=70
)
sns.regplot(
    x="StudyHours",
    y="Marks",
    data=df,
    scatter=False
)
plt.title("Study Hours vs Marks by Result")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.legend(title="Result")
```

```python
sns.pairplot(
    df[
        [
        "StudyHours",
        "Attendance",
        "PrevMarks",
        "Marks"
        ]
        ]
)
plt.show()
```

```python
correlation = df[
    [
        "StudyHours",
        "Attendance",
        "PrevMarks",
        "Marks"
    ]
    ].corr()
print("\nCorrelation Matrix:")
print(correlation)
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(
    correlation,
    annot=True
)
plt.title("Correlation Heatmap")
plt.show()
```
